```python
import numpy as np
import matplotlib.pyplot as plt

tenors = np.array([0.25, 0.5, 1, 2, 5, 10])

# Synthetic yield curves
usd_curve = np.array([5.25, 5.15, 4.90, 4.50, 4.10, 4.00])
jpy_curve = np.array([0.20, 0.25, 0.35, 0.55, 0.85, 1.10])
eur_curve = np.array([3.75, 3.60, 3.35, 3.00, 2.70, 2.60])

plt.figure(figsize=(10, 5))
plt.plot(tenors, usd_curve, marker="o", label="USD curve")
plt.plot(tenors, jpy_curve, marker="o", label="JPY curve")
plt.plot(tenors, eur_curve, marker="o", label="EUR curve")

plt.title("Synthetic G10 Yield Curves")
plt.xlabel("Tenor, years")
plt.ylabel("Yield, %")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
usd_rate = 5.00
jpy_rate = 0.50

carry = usd_rate - jpy_rate

fx_moves = np.linspace(-15, 15, 100)  # % change in USDJPY
total_return = carry + fx_moves

plt.figure(figsize=(10, 5))
plt.plot(fx_moves, total_return, label="Total return")
plt.axhline(0, color="black", linewidth=1)
plt.axvline(-carry, color="red", linestyle="--", label="Break-even FX move")

plt.title("Long USD / Short JPY Carry Trade")
plt.xlabel("USDJPY move, %")
plt.ylabel("Total return, %")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print(f"Annual carry: {carry:.2f}%")
print(f"Break-even USDJPY move: {-carry:.2f}%")
```

```python
scenarios = {
    "Current": (5.00, 0.50),
    "Fed cuts": (3.50, 0.50),
    "BoJ hikes": (5.00, 1.50),
    "Both move against carry": (3.50, 1.50),
}

labels = []
carries = []

for name, (usd, jpy) in scenarios.items():
    labels.append(name)
    carries.append(usd - jpy)

plt.figure(figsize=(9, 5))
plt.bar(labels, carries)
plt.title("USD/JPY Carry Under Rate Scenarios")
plt.ylabel("Carry, %")
plt.axhline(0, color="black", linewidth=1)
plt.grid(axis="y", alpha=0.3)
plt.show()
```
